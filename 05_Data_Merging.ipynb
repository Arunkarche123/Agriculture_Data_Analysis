```python
import pandas as pd
```

```python
dataset1 = pd.read_csv("Dataset1_Cleaned_AGMARKNET.csv")

dataset2 = pd.read_csv("Dataset2_Cleaned_Market_Locations.csv")

dataset3 = pd.read_csv("Dataset3_Cleaned_Weather.csv")
```

```python
print("Dataset 1 :", dataset1.shape)

print("Dataset 2 :", dataset2.shape)

print("Dataset 3 :", dataset3.shape)
```

```python
dataset1.head()
```

```python
dataset2.head()
```

```python
dataset3.head()
```

```python
merged_df = dataset1.merge(
    dataset2[
        [
            "state",
            "district",
            "market",
            "latitude",
            "longitude"
        ]
    ],
    on=["state", "district", "market"],
    how="left",
    validate="many_to_one"
)

print(merged_df.shape)
```

```python
print(merged_df[["latitude", "longitude"]].isnull().sum())
```

```python
print(dataset3.columns.tolist())
```

```python
print(dataset2.columns.tolist())
```

```python
final_df = merged_df.merge(

    dataset3[
        [
            "market",
            "latitude",
            "longitude",
            "temperature",
            "humidity",
            "rainfall",
            "wind_speed",
            "weather_code"
        ]
    ],

    on=[
        "market",
        "latitude",
        "longitude"
    ],

    how="left",
    validate="many_to_one"

)

print(final_df.shape)
```

```python
print(final_df.isnull().sum())
```

```python
print(final_df.duplicated().sum())
```

```python
final_df[final_df["humidity"].isnull()][
    ["state","district","market","latitude","longitude"]
]
```

```python
print("Rows:", final_df.shape[0])
print("Columns:", final_df.shape[1])

assert final_df.shape[0] == 9049, "Row count mismatch!"

assert final_df.duplicated().sum() == 0, "Duplicate rows found!"

assert final_df[
    ["latitude","longitude","temperature"]
].isnull().sum().sum() == 0, "Missing values detected!"

print("✅ Final Dataset Validation Passed")
```

```python
final_df.to_csv(
    "Master_Agriculture_Mandi_Dataset.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Master Dataset Saved Successfully!")
```
