## 07 Feature Engineering And EDA

```python
df1.shape
```

```python
# 1. Create Price Spread
df1["price_spread"] = df1["max_price"] - df1["min_price"]
```

```python
# Preview
df1[["min_price", "max_price", "price_spread"]].head()
```

```python
# 2. Create Average Price
df1["average_price"] = (df1["min_price"] + df1["max_price"]) / 2
```

```python
# Preview
df1[["min_price", "max_price", "average_price"]].head()
```

```python
# 3. Create Price Category

def price_category(price):
    if price < 2000:
        return "Low"
    elif price <= 5000:
        return "Medium"
    else:
        return "High"

df1["price_category"] = df1["average_price"].apply(price_category)
```

```python
# Preview
df1[["average_price", "price_category"]].head()
```

```python
# 4. Create Temperature Category

def temperature_category(temp):
    if temp < 20:
        return "Cold"
    elif temp <= 30:
        return "Moderate"
    else:
        return "Hot"

df1["temperature_category"] = df1["temperature"].apply(temperature_category)
```

```python
# Preview
df1[["temperature", "temperature_category"]].head()
```

```python
# 5. Create Humidity Category

def humidity_category(humidity):
    if humidity < 40:
        return "Low"
    elif humidity <= 70:
        return "Medium"
    else:
        return "High"

df1["humidity_category"] = df["humidity"].apply(humidity_category)
```

```python
# Preview
df1[["humidity", "humidity_category"]].head()
```

```python
# 6. Create Rainfall Category

def rainfall_category(rainfall):
    if rainfall == 0:
        return "No Rain"
    elif rainfall <= 20:
        return "Light Rain"
    else:
        return "Heavy Rain"

df1["rainfall_category"] = df["rainfall"].apply(rainfall_category)
```

```python
# Preview
df1[["rainfall", "rainfall_category"]].head()
```

```python
# Display newly created feature columns

new_features = [
    "price_spread",
    "average_price",
    "price_category",
    "temperature_category",
    "humidity_category",
    "rainfall_category"
]

df1[new_features].head()
```
